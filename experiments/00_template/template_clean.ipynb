```python
import sys
import os
from pathlib import Path

# Add project root to path
project_root = Path.cwd().parent.parent
sys.path.insert(0, str(project_root))
sys.path.insert(0, str(project_root / 'src'))

print(f"Project root: {project_root}")
print(f"Working directory: {Path.cwd()}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader

# Project imports - import your model here
from models.unet import UNet  # or ResNetUNet, etc.
from data_loader.dataset import RetinaDataset, RetinaDatasetTest

# Import utility functions - these eliminate code duplication!
from experiments.utils import (
    train_model,           # Handles full training loop
    test_model,            # Tests model on test set
    plot_training_curves,  # Plots loss and dice curves
    visualize_predictions, # Creates prediction visualizations
    visualize_data_samples,# Shows data samples
    save_config,          # Saves config to JSON
    save_history,         # Saves training history
    save_test_results,    # Saves test results
    print_test_results,   # Prints formatted test results
    print_model_info,     # Prints model parameter count
)

# Check GPU availability
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
```

## 2️⃣ Configuration

**This is where your experiments differ!** Just change the config, and the utility functions handle everything else.

```python
# Experiment configuration - CUSTOMIZE THIS FOR YOUR EXPERIMENT
CONFIG = {
    # Model
    'model': 'your_model_name',
    'base_features': 64,  # or other model-specific params
    
    # Data
    'data_dir': '../../datasets/REFUGE',
    'train_csv': '../../datasets/REFUGE/REFUGETrain.csv',
    'val_csv': '../../datasets/REFUGE/REFUGE1Val.csv',
    'test_csv': '../../datasets/REFUGE/REFUGE1Test.csv',
    'use_cropped': True,
    'use_clahe': False,  # Change to True for CLAHE
    
    # Training
    'epochs': 50,
    'batch_size': 8,
    'learning_rate': 1e-4,
    'weight_decay': 1e-4,
    'num_workers': 4,
    
    # Output
    'output_dir': './results',
}

# Create output directory
os.makedirs(CONFIG['output_dir'], exist_ok=True)
os.makedirs(f"{CONFIG['output_dir']}/visualizations", exist_ok=True)

# Save configuration
save_config(CONFIG, f"{CONFIG['output_dir']}/config.json")
```

## 3️⃣ Data Loading

```python
# Create datasets
train_dataset = RetinaDataset(
    csv_file=CONFIG['train_csv'],
    root_dir=CONFIG['data_dir'],
    use_cropped=CONFIG['use_cropped'],
    use_clahe=CONFIG['use_clahe'],
)

val_dataset = RetinaDataset(
    csv_file=CONFIG['val_csv'],
    root_dir=CONFIG['data_dir'],
    use_cropped=CONFIG['use_cropped'],
    use_clahe=CONFIG['use_clahe'],
)

test_dataset = RetinaDatasetTest(
    csv_file=CONFIG['test_csv'],
    root_dir=CONFIG['data_dir'],
    use_cropped=CONFIG['use_cropped'],
    use_clahe=CONFIG['use_clahe'],
)

print(f"✅ Train: {len(train_dataset)}, Val: {len(val_dataset)}, Test: {len(test_dataset)}")
```

```python
# Create data loaders
train_loader = DataLoader(
    train_dataset, batch_size=CONFIG['batch_size'], shuffle=True,
    num_workers=CONFIG['num_workers'], pin_memory=True
)

val_loader = DataLoader(
    val_dataset, batch_size=CONFIG['batch_size'], shuffle=False,
    num_workers=CONFIG['num_workers'], pin_memory=True
)

test_loader = DataLoader(
    test_dataset, batch_size=CONFIG['batch_size'], shuffle=False,
    num_workers=CONFIG['num_workers'], pin_memory=True
)

print("✅ Data loaders created")
```

```python
# Visualize sample data
visualize_data_samples(
    train_dataset,
    output_path=f"{CONFIG['output_dir']}/data_samples.png",
    num_samples=2
)
```

## 4️⃣ Model Definition

```python
# Create model - CUSTOMIZE THIS FOR YOUR MODEL
model = UNet(
    n_channels=3,
    n_classes=2,
    base_features=CONFIG['base_features']
).to(device)

# Print model info using utility function
print_model_info(model, title=f"✅ Model created: {CONFIG['model']}")
```

```python
# Define loss function and optimizer
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(
    model.parameters(),
    lr=CONFIG['learning_rate'],
    weight_decay=CONFIG['weight_decay']
)

print("✅ Loss and optimizer defined")
```

## 5️⃣ Training

**No need to write training loop code!** The `train_model()` utility function handles everything.

```python
# Train model using utility function
history = train_model(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    criterion=criterion,
    optimizer=optimizer,
    config=CONFIG,
    device=device,
    output_dir=CONFIG['output_dir']
)

# Save training history
save_history(history, f"{CONFIG['output_dir']}/training_history.json")
```

```python
# Plot training curves
plot_training_curves(
    history=history,
    output_path=f"{CONFIG['output_dir']}/training_curves.png"
)
```

## 6️⃣ Testing

```python
# Load best model
checkpoint = torch.load(f"{CONFIG['output_dir']}/best_model.pth")
model.load_state_dict(checkpoint['model_state_dict'])
print("✅ Best model loaded")
```

```python
# Test model using utility function
test_results, predictions, masks = test_model(
    model=model,
    test_loader=test_loader,
    device=device
)

# Print and save results
print_test_results(test_results)
save_test_results(test_results, f"{CONFIG['output_dir']}/test_results.json")
```

```python
# Visualize predictions
visualize_predictions(
    test_dataset=test_dataset,
    predictions=predictions,
    masks=masks,
    output_path=f"{CONFIG['output_dir']}/visualizations/test_predictions.png",
    num_samples=4,
    seed=42
)
```
